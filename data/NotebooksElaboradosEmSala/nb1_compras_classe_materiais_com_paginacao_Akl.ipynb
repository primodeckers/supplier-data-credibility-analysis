Importar bibliotecas

```python
import requests
import pandas as pd
```

URL da API que retornará os dados. Desta vez, vamos utilizar uma chamada que retorna mais dados para explorarmos a paginação. No caso utilizaremos a consulta a classes de material.

```python
serv_url_classe = "https://dadosabertos.compras.gov.br/modulo-material/2_consultarClasseMaterial"
```

```python
pagina = 1
resultado_classe=[]
```

```python
while True:
  response = requests.get(serv_url_classe, params={'pagina': pagina})
  if response.status_code == 200:
    json_response = response.json()
    resultado_classe.extend(json_response["resultado"])
    print(f"Página {pagina} baixada com sucesso!")
    paginas_restantes = json_response["paginasRestantes"]
    print(f"Páginas restantes: {paginas_restantes}")
    if paginas_restantes == 0:
      break
    pagina += 1
  else:
      print("Erro:", response.text)
      break
```

```python
df_classe = pd.DataFrame(resultado_classe)
display(df_classe.head(10))
```
