Importando as bibliotecas que serão exploradas

```python
import requests
import pandas as pd
```

Endereço da API (Swagger)

https://dadosabertos.compras.gov.br

```python
serv_url_grupo = "https://dadosabertos.compras.gov.br/modulo-material/1_consultarGrupoMaterial"
```

Executando a chamada REST

```python
response = requests.get(serv_url_grupo)
```

```python
response = requests.get(serv_url_grupo)
if response.status_code == 200: #Sucesso
    resultado_grupo = response.json()
    print(response.json())
else:
    print("Erro: ", response.text)
```

Armazenando o resultado em uma variável (resultado_grupo")

```python
resultado_grupo = resultado_grupo["resultado"]
print(resultado_grupo)
```

Importando o resultado em um dataFrame

```python
df_grupo = pd.DataFrame(resultado_grupo)
df_grupo.head()
```
