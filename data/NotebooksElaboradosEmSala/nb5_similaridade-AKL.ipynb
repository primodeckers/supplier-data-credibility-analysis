Vamos usar em casos reais. Na questão do CNAE.

```python
df_cnae_fornecedores = pd.read_pickle(os.path.join(DATA_PATH, "cnaes_transf.pkl"))
df_cnae_fornecedores
```

```python
import unicodedata
def padroniza_str(str):
  nfd = unicodedata.normalize("NFD", str)
  return nfd.encode("ascii", "ignore").decode("utf-8").upper()
```

```python
df_cnae = pd.read_csv(os.path.join(DATA_PATH,'CNAERFB.CSV'),
                      sep=';',
                      encoding='latin-1',
                      names=['id_cnae', 'desc_cnae'])
df_cnae.desc_cnae = pd.Series(map(padroniza_str, df_cnae.desc_cnae))
df_cnae
```

```python
df_merge_cnae = df_cnae_fornecedores.merge(df_cnae, how='left',left_on='id', right_on='id_cnae')
condicao = df_merge_cnae.descricao == df_merge_cnae.desc_cnae
df_merge_cnae[condicao]
```

```python
df_merge_cnae[~condicao]
```

```python
qtd_ok = len(df_merge_cnae[condicao])
print(qtd_ok)
qtd_ok = len(df_merge_cnae[~condicao])
print(qtd_ok)
```

```python
df_diff = df_merge_cnae[~condicao][['descricao', 'desc_cnae']]
df_diff
```

```python
distance.nlevenshtein(df_diff.descricao, df_diff.desc_cnae, method=1)
```

```python
def lev_method2(str1, str2):
    x = distance.nlevenshtein(str1, str2, method=2)
    return x
```

```python
df_diff['dist_lev'] = list(map(lev_method2, df_diff.descricao,df_diff.desc_cnae))
df_diff.sort_values(by='dist_lev',ascending=False)
```
